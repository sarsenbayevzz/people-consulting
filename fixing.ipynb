```python
import pandas as pd
```

```python
r = pd.read_csv('respondents.csv')
```

```python
city_map = {
    'Другое: Шымкент': 'Шымкент',
    'Другое: Актобе': 'Актобе',
    'Другое: Ақтөбе': 'Актобе',
    'Другое: семей': 'Семей',
    'Другое: Семей': 'Семей',
    'Другое: Санкт-Петербург': 'Санкт-Петербург',
    'Другое: Москва': 'Москва',
    'Другое: Кокшетау': 'Кокшетау',
    'Другое: Петропавловск': 'Петропавловск',
    'Другое: Павлодар': 'Павлодар',
    'Другое: Ташкент': 'Ташкент',
    'Другое: Актау': 'Актау',
    'Другое: Тараз': 'Тараз',
    'Другое: Каскелен': 'Каскелен',
}

r['city'] = r['city'].replace(city_map)
r.city.unique()
```

```python
r.loc[r['current_industry'].str.startswith('Другое:', na=False), 'current_industry'] = 'Другое'
r.current_industry.unique()
```

```python
r.to_csv('respondents.csv', index=False)
```
