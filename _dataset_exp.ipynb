```python
import pandas as pd

# ====== ЗАГРУЗКА ДАННЫХ ======
r = pd.read_csv('respondents.csv')
c26 = pd.read_csv('company_2026.csv')
c25 = pd.read_csv('company_2025.csv')
f = pd.read_csv('factors.csv')
y = pd.read_csv('yoy_compare.csv')
fcv = pd.read_csv('factor_categories_viz.csv')
ftv = pd.read_csv('factor_tokens_viz.csv')
scv = pd.read_csv('survey_categories_viz.csv')
stv = pd.read_csv('survey_tokens_viz.csv')
tmc = pd.read_csv('top_missed_companies.csv')

dfs = {
    "respondents": r,
    "company_2026": c26,
    "company_2025": c25,
    "factors": f,
    "yoy_compare": y,
    "fcv": fcv,
    "ftv": ftv,
    "scv": scv,
    "stv": stv,
    "tmc": tmc
}

# ====== ФУНКЦИЯ EDA ======
def full_eda(df, name):
    print(f"\n{'='*60}")
    print(f"DATASET: {name}")
    print(f"{'='*60}")
    
    # 1. Shape
    print("\n📐 Shape:")
    print(df.shape)
    
    # 2. Columns + types
    print("\n📊 Columns & Types:")
    print(df.dtypes)
    
    # 3. Missing values
    print("\n❗ Missing values:")
    missing = df.isnull().sum()
    print(missing[missing > 0] if missing.sum() > 0 else "No missing values")
    
    # 4. Duplicates
    print("\n🔁 Duplicates:")
    print(df.duplicated().sum())
    
    # 5. Numeric summary
    print("\n📈 Numeric Summary:")
    print(df.describe())
    
    # 6. Categorical columns
    cat_cols = df.select_dtypes(include='object').columns
    
    print("\n🔤 Categorical Columns:")
    print(list(cat_cols))
    
    # 7. Unique values + top values
    for col in cat_cols:
        print(f"\n--- {col} ---")
        print("Unique count:", df[col].nunique())
        
        # если слишком много значений — покажем топ
        if df[col].nunique() > 20:
            print("Top 10 values:")
            print(df[col].value_counts().head(10))
        else:
            print("All unique values:")
            print(df[col].unique())
    
    # 8. Correlation (если есть числа)
    if len(df.select_dtypes(include='number').columns) > 1:
        print("\n🔗 Correlation:")
        print(df.corr(numeric_only=True))


# ====== ЗАПУСК ДЛЯ ВСЕХ ======
for name, df in dfs.items():
    full_eda(df, name)
```
